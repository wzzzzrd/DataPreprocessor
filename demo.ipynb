```python
import pandas as pd
import seaborn as sns
from preprocessor import DataPreprocessor
from pprint import pprint

df_raw = sns.load_dataset('titanic')

columns = ['survived', 'pclass', 'sex', 'age', 'sibsp', 'fare', 'embarked', 'deck']
df = df_raw[columns].copy()

print(f"Количество строк и столбцов: {df.shape}")

display(df.head(20))

preprocessor = DataPreprocessor(df)

clean_df = preprocessor.fit_transform(threshold=0.5, strategy='median', method='minmax')

print("Обработанный датасет:")
display(clean_df.head(20))
print("Журнал преобразований:")
pprint(preprocessor.history, sort_dicts=False)
```
